```python
candidates = {
    1: {"name": "Candidate A", "votes": 0},
    2: {"name": "Candidate B", "votes": 0},
    3: {"name": "Candidate C", "votes": 0}
}

voters = {}



def register_voter():
    voter_id = input("Enter Voter ID: ")

    if voter_id in voters:
        print("❌ Voter already registered!")
        return

    name = input("Enter Voter Name: ")

    voters[voter_id] = {
        "name": name,
        "has_voted": False
    }

    print("✅ Voter registered successfully!")



def display_candidates():
    print("\n========== CANDIDATES ==========")

    for number, candidate in candidates.items():
        print(f"{number}. {candidate['name']}")

    print("================================")



def cast_vote():
    voter_id = input("Enter your Voter ID: ")

    if voter_id not in voters:
        print("❌ Voter not registered!")
        return

    if voters[voter_id]["has_voted"]:
        print("❌ You have already voted!")
        return

    display_candidates()

    try:
        choice = int(input("Enter candidate number: "))

        if choice not in candidates:
            print("❌ Invalid candidate!")
            return

        candidates[choice]["votes"] += 1


        voters[voter_id]["has_voted"] = True

        print("✅ Your vote has been successfully recorded!")

    except ValueError:
        print("❌ Please enter a valid number.")



def display_results():
    print("\n========== VOTING RESULTS ==========")

    total_votes = 0

    for number, candidate in candidates.items():
        print(
            f"{candidate['name']} : "
            f"{candidate['votes']} vote(s)"
        )
        total_votes += candidate["votes"]

    print("------------------------------------")
    print("Total Votes:", total_votes)


    if total_votes > 0:
        max_votes = max(
            candidate["votes"]
            for candidate in candidates.values()
        )

        winners = [
            candidate["name"]
            for candidate in candidates.values()
            if candidate["votes"] == max_votes
        ]

        if len(winners) == 1:
            print("🏆 Winner:", winners[0])
        else:
            print("🤝 Result: Tie between", ", ".join(winners))

    print("====================================")


def main():
    while True:
        print("\n")
        print("======================================")
        print("       ONLINE VOTING SYSTEM")
        print("======================================")
        print("1. Register Voter")
        print("2. View Candidates")
        print("3. Cast Vote")
        print("4. View Results")
        print("5. Exit")
        print("======================================")

        choice = input("Enter your choice: ")

        if choice == "1":
            register_voter()

        elif choice == "2":
            display_candidates()

        elif choice == "3":
            cast_vote()

        elif choice == "4":
            display_results()

        elif choice == "5":
            print("Thank you for using the Online Voting System!")
            break

        else:
            print("❌ Invalid choice. Please try again.")



main()
```
